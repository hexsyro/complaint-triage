## 📓 Notebook 02: Vectorization — CountVectorizer vs. TF-IDF

**Skills practiced:** `CountVectorizer`, `TfidfVectorizer`, reading the vocabulary, `stop_words`, `ngram_range`, comparing weights on the same few examples.

**Goal:** Fit both vectorizers on 5 sample complaints, print the actual numbers, and see why TF-IDF down-weights words that appear everywhere.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

import pandas as pd
```

```python
df=pd.read_csv('../data/processed/complaints_sample.csv')
```

```python
vectorizer = CountVectorizer()
sparse_matrix = vectorizer.fit_transform(df['Issue'].head(5))
words = vectorizer.get_feature_names_out()
dense = sparse_matrix.toarray()
print(words)
print(dense)
```

```python
Tvectorizer=TfidfVectorizer()
sparse_matrix = Tvectorizer.fit_transform(df['Issue'].head(5))
words = Tvectorizer.get_feature_names_out()
dense = sparse_matrix.toarray()
print(words)
print(dense)
```

```python
Tvectorizer=TfidfVectorizer(stop_words='english')
sparse_matrix = Tvectorizer.fit_transform(df['Issue'].head(5))
words = Tvectorizer.get_feature_names_out()
dense = sparse_matrix.toarray()
print(words)
print(dense)
```

```python
Tvectorizer=TfidfVectorizer(stop_words='english', ngram_range=(1,2))
sparse_matrix = Tvectorizer.fit_transform(df['Issue'].head(5))
words = Tvectorizer.get_feature_names_out()
dense = sparse_matrix.toarray()
print(words)
print(dense)
```
